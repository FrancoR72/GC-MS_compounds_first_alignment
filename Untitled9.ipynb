```python
# @title
# ============================================================
# Validazione Miscele ricostruite con Algoritmo Genetico
# ------------------------------------------------------------
# CELLA 1 — Generazione Target unitari
#
# OPZIONE 1: Target unitario, da Mono-Origine unitari
# OPZIONE 2: Target unitario, da Mono-Origine multi-batch
# OPZIONE 3: Target probabilistico, da Mono-Origine unitari
# OPZIONE 4: Target probabilistico, da Mono-Origine multi-batch
#            (Opzioni 3/4: stessa processazione di generazione delle
#             Opzioni 1/2; al termine passa automaticamente alla
#             sezione "Gestione del Target Probabilistico", che per
#             ora si limita al caricamento e alla lettura del file
#             Target probabilistico)
#
# OUTPUT:
#   - N Target <= 25: un foglio per Target in ciascuno dei 2 file xlsx
#   - N Target  > 25: righe impilate in un unico foglio per ciascun file,
#     + colonna Target_ID, + riga vuota di separazione tra gruppi
#     (solo nel file Mono-Origine), + istogrammi (bin 1%) e xlsx
#     statistiche di distribuzione per origine, mostrati inline
#     e scaricabili separatamente
#   - N Target  > 100: salva in `state` i dati necessari alla cella
#     "Verifica Algoritmo Genetico"
#   - SEMPRE (qualunque N): grafico radar/spider con un profilo per
#     formulazione, mostrato inline e scaricabile; dopo la prima
#     visualizzazione sono disponibili controlli per autoscaling,
#     colore e spessore delle linee, con aggiornamento del grafico
#     mostrato E del file scaricabile
#   - Download tramite pulsanti manuali (nessun download automatico)
# ============================================================

!pip -q install pandas openpyxl numpy ipywidgets matplotlib scipy

import io, os, re, string, unicodedata, zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis
from typing import Dict, List, Tuple, Optional
import ipywidgets as widgets
from IPython.display import display, clear_output

try:
    from google.colab import files
    IN_COLAB = True
except Exception:
    IN_COLAB = False
    print("⚠️ Questo notebook non è in Google Colab: il download manuale non funzionerà.")

# ---------- Costanti struttura file (allineate a MMv3_11) ----------
HEADER_ROW_INDEX = 0
START_COL_INDEX = 3
BATCH_COL_INDEX = 2
EPS = 1e-9

EXCLUDED_ORIGINS = {"REF", "QC"}
STACK_THRESHOLD = 25
VERIFICATION_THRESHOLD = 100  # soglia per abilitare la cella "Verifica Algoritmo Genetico"
HIST_BIN_WIDTH_PCT = 1  # larghezza bin istogrammi, in punti percentuali
RADAR_TICK_STEP = 10    # passo delle tacche sugli assi del radar chart, in punti percentuali

DEFAULT_MINMAX: Dict[str, Tuple[float, float]] = {
    "IVORY COAST": (25.0, 50.0),
    "NIGERIA":     (25.0, 50.0),
    "GHANA":       (0.0, 25.0),
    "EC ARRIBA":   (0.0, 20.0),
    "ARRIBA":      (0.0, 20.0),
    "EC CCN51":    (10.0, 50.0),
    "CCN51":       (10.0, 50.0),
}
FALLBACK_MINMAX = (0.0, 100.0)


# ============================================================
# Utility comuni
# ============================================================

def _strip_accents(text: str) -> str:
    text = unicodedata.normalize("NFKD", str(text))
    return "".join(ch for ch in text if not unicodedata.combining(ch))

def normalize_origin_key(value) -> str:
    s = _strip_accents(value).upper().strip()
    s = re.sub(r"[^A-Z0-9]+", " ", s).strip()
    s = re.sub(r"\s+", " ", s)
    return s

_DEFAULT_MINMAX_NORM = {normalize_origin_key(k): v for k, v in DEFAULT_MINMAX.items()}

def get_default_minmax(origin_name: str) -> Tuple[float, float]:
    return _DEFAULT_MINMAX_NORM.get(normalize_origin_key(origin_name), FALLBACK_MINMAX)

def is_excluded_origin(value) -> bool:
    return normalize_origin_key(value) in EXCLUDED_ORIGINS

def detect_origin_column(df: pd.DataFrame) -> str:
    strong_keys = {"ORIGINE", "ORIGINI", "ORIGIN", "ORIGINS", "CLASSE", "CLASSI",
                   "CLASS", "CLASSES", "PROVENIENZA", "PROVENANCE", "COUNTRY", "PAESE"}
    for c in df.columns:
        if normalize_origin_key(c) in strong_keys:
            return c
    for c in df.columns:
        ck = normalize_origin_key(c)
        if any(tok in ck for tok in ["ORIGIN", "CLASS", "PROVEN"]) or ck in {"COUNTRY", "PAESE"}:
            return c
    raise ValueError("Non trovo una colonna riconducibile a 'origine/classe' nel file caricato.")

def detect_batch_column(df: pd.DataFrame, origin_col: str) -> str:
    strong_keys = {"BATCH", "BATCHID", "LOTTO", "LOT", "LOTID"}
    for c in df.columns:
        if c == origin_col:
            continue
        if normalize_origin_key(c) in strong_keys:
            return c
    for c in df.columns:
        if c == origin_col:
            continue
        if "BATCH" in normalize_origin_key(c) or "LOTTO" in normalize_origin_key(c):
            return c
    if BATCH_COL_INDEX < df.shape[1]:
        print(f"ℹ️ Colonna Batch non riconosciuta per nome: uso la colonna posizionale "
              f"'{df.columns[BATCH_COL_INDEX]}' (indice {BATCH_COL_INDEX}).")
        return df.columns[BATCH_COL_INDEX]
    raise ValueError("Non trovo una colonna Batch nel file caricato.")

def target_letter(i: int) -> str:
    letters = string.ascii_uppercase
    label = ""
    n = i
    while True:
        n, r = divmod(n, 26)
        label = letters[r] + label
        if n == 0:
            break
        n -= 1
    return label


# ============================================================
# Caricamento file (comune alle quattro opzioni)
# ============================================================

def load_xlsx_from_upload(prompt: str) -> pd.DataFrame:
    if not IN_COLAB:
        raise RuntimeError("Upload via browser disponibile solo in Google Colab.")
    print(prompt)
    uploaded = files.upload()
    fname = list(uploaded.keys())[0]
    df = pd.read_excel(io.BytesIO(uploaded[fname]), sheet_name=0, header=HEADER_ROW_INDEX)
    print(f"✅ File caricato: {fname} — {df.shape[0]} righe, {df.shape[1]} colonne.")
    return df


# ============================================================
# Campionamento pesi entro i limiti min/max, somma = 100%
# ============================================================

def sample_bounded_composition(mins: np.ndarray, maxs: np.ndarray,
                                total: float = 100.0, rng: np.random.Generator = None,
                                max_iter: int = 500) -> np.ndarray:
    rng = rng or np.random.default_rng()
    n = len(mins)

    if mins.sum() - EPS > total:
        raise ValueError(
            f"Vincoli infattibili: la somma dei minimi ({mins.sum():.1f}%) supera il totale richiesto ({total}%)."
        )
    if maxs.sum() + EPS < total:
        raise ValueError(
            f"Vincoli infattibili: la somma dei massimi ({maxs.sum():.1f}%) è inferiore al totale richiesto ({total}%)."
        )

    w = mins.copy().astype(float)
    capacity = (maxs - mins).astype(float)
    remaining = total - w.sum()

    it = 0
    while remaining > EPS and it < max_iter:
        active = capacity > EPS
        if not np.any(active):
            break
        shares = rng.dirichlet(np.ones(active.sum()))
        add = np.zeros(n)
        add[active] = shares * remaining
        actual_add = np.minimum(add, capacity)
        w += actual_add
        capacity -= actual_add
        remaining = total - w.sum()
        it += 1

    if remaining > 1e-3:
        idx = np.argmax(capacity)
        w[idx] += min(remaining, capacity[idx])

    return w


# ============================================================
# OPZIONI 1 / 3 — costruzione fogli per un Target (mono-origine unitaria)
# ============================================================

def build_target_sheets_opt1(mono_df: pd.DataFrame, var_cols: List[str],
                              w_true_pct: np.ndarray) -> Tuple[pd.DataFrame, pd.DataFrame, str]:
    cols_before = list(mono_df.columns[:START_COL_INDEX])
    cols_vars = list(mono_df.columns[START_COL_INDEX:])

    sheet1 = mono_df[cols_before].copy()
    sheet1.insert(len(cols_before), "Perc_Target_%", np.round(w_true_pct, 4))
    for c in cols_vars:
        sheet1[c] = mono_df[c].values

    X = mono_df[var_cols].to_numpy(dtype=float)
    w_frac = (w_true_pct / 100.0).reshape(1, -1)
    target_profile = (w_frac @ X).flatten()
    sheet2 = pd.DataFrame([target_profile], columns=var_cols)

    origin_field_name = cols_before[-1] if cols_before else mono_df.columns[0]
    for c in cols_before:
        if normalize_origin_key(c) in {"ORIGINE", "ORIGIN"} or "ORIGIN" in normalize_origin_key(c):
            origin_field_name = c
            break

    return sheet1, sheet2, origin_field_name


# ============================================================
# OPZIONI 2 / 4 — costruzione fogli per un Target (mono-origine multi-batch)
# ============================================================

def build_target_sheets_opt2(multi_df: pd.DataFrame, origin_col: str, batch_col: str,
                              var_cols: List[str], origin_names: List[str],
                              w_true_pct: np.ndarray,
                              rng: np.random.Generator) -> Tuple[pd.DataFrame, pd.DataFrame, str]:

    selected_rows = []
    selected_meta = []

    for name in origin_names:
        subset = multi_df[multi_df[origin_col].astype(str) == name]
        batches_available = subset[batch_col].dropna().unique().tolist()
        if len(batches_available) == 0:
            raise ValueError(f"Origine '{name}': nessun batch disponibile dopo il filtro REF/QC.")

        chosen_batch = batches_available[rng.integers(0, len(batches_available))]
        replicas = subset[subset[batch_col] == chosen_batch]
        if replicas.shape[0] == 0:
            raise ValueError(f"Origine '{name}', batch '{chosen_batch}': nessuna replica trovata.")

        chosen_idx = replicas.index[rng.integers(0, replicas.shape[0])]
        chosen_row = multi_df.loc[chosen_idx]

        selected_rows.append(chosen_row[var_cols].to_numpy(dtype=float))
        selected_meta.append({
            "Origine": name,
            "Batch_selezionato": chosen_batch,
            "Indice_riga_originale": chosen_idx,
            "N_repliche_disponibili_nel_batch": replicas.shape[0],
        })

    X_selected = np.vstack(selected_rows)

    sheet1 = pd.DataFrame(selected_meta)
    sheet1.insert(len(sheet1.columns), "Perc_Target_%", np.round(w_true_pct, 4))
    for j, c in enumerate(var_cols):
        sheet1[c] = X_selected[:, j]

    w_frac = (w_true_pct / 100.0).reshape(1, -1)
    target_profile = (w_frac @ X_selected).flatten()
    sheet2 = pd.DataFrame([target_profile], columns=var_cols)

    return sheet1, sheet2, "Origine"


# ============================================================
# Statistiche di distribuzione + istogrammi per origine
# (mostrati inline; xlsx statistiche e zip istogrammi separati,
# scaricabili come file distinti)
# ============================================================

def build_and_show_distributions(pct_by_origin: Dict[str, List[float]],
                                  hist_zip_path: str, stats_xlsx_path: str) -> pd.DataFrame:
    stats_rows = []
    png_paths = []
    n_bins = int(round(100 / HIST_BIN_WIDTH_PCT))

    print("\n--- Istogrammi di distribuzione % per origine (bin da "
          f"{HIST_BIN_WIDTH_PCT} punto percentuale) ---")

    for origin_name, values in pct_by_origin.items():
        arr = np.array(values, dtype=float)
        mean_v = float(np.mean(arr))
        median_v = float(np.median(arr))
        std_v = float(np.std(arr, ddof=1)) if len(arr) > 1 else 0.0
        skew_v = float(skew(arr)) if len(arr) > 2 else np.nan
        kurt_v = float(kurtosis(arr)) if len(arr) > 3 else np.nan

        stats_rows.append({
            "Origine": origin_name,
            "N_target": len(arr),
            "Media_%": mean_v,
            "Mediana_%": median_v,
            "Dev_Std_%": std_v,
            "Curtosi": kurt_v,
            "Skewness": skew_v,
        })

        fig, ax = plt.subplots(figsize=(6.5, 4.2))
        ax.hist(arr, bins=n_bins, range=(0, 100), edgecolor="black", alpha=0.75)
        ax.set_xlim(0, 100)
        ax.set_xlabel("% di impiego nel Target")
        ax.set_ylabel("Frequenza")
        ax.set_title(f"Distribuzione % impiego — {origin_name}")

        stats_text = (
            f"Media: {mean_v:.2f}%\n"
            f"Mediana: {median_v:.2f}%\n"
            f"Dev.Std: {std_v:.2f}%\n"
            f"Curtosi: {kurt_v:.2f}\n"
            f"Skewness: {skew_v:.2f}"
        )
        ax.text(0.98, 0.97, stats_text, transform=ax.transAxes,
                fontsize=9, va="top", ha="right",
                bbox=dict(boxstyle="round", facecolor="white", alpha=0.85))

        fig.tight_layout()
        png_name = f"distribuzione_{re.sub(r'[^A-Za-z0-9]+', '_', origin_name)}.png"
        fig.savefig(png_name, dpi=150)

        plt.show()
        plt.close(fig)

        png_paths.append(png_name)

    stats_df = pd.DataFrame(stats_rows)

    print("\n--- Tabella riepilogativa statistiche di distribuzione ---")
    display(stats_df.round(3))

    stats_df.to_excel(stats_xlsx_path, index=False)

    with zipfile.ZipFile(hist_zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
        for p in png_paths:
            zf.write(p)
            os.remove(p)

    return stats_df


# ============================================================
# Radar/spider chart — un profilo per formulazione generata
# ============================================================

def render_radar_chart(origin_names: List[str], all_weights_pct: List[np.ndarray],
                        radar_path: str, line_color: str = "lightgray",
                        line_width: float = 0.6, autoscale: bool = False):
    n_axes = len(origin_names)
    angles = [n / float(n_axes) * 2 * np.pi for n in range(n_axes)]
    angles += angles[:1]

    if autoscale:
        overall_max = max(float(np.max(w)) for w in all_weights_pct)
        axis_max = max(RADAR_TICK_STEP, np.ceil(overall_max / RADAR_TICK_STEP) * RADAR_TICK_STEP)
    else:
        axis_max = 100.0

    fig = plt.figure(figsize=(7, 7))
    ax = plt.subplot(111, polar=True)
    ax.set_theta_offset(np.pi / 2)
    ax.set_theta_direction(-1)

    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(origin_names, fontsize=10)

    ticks = list(np.arange(0, axis_max + EPS, RADAR_TICK_STEP))
    ax.set_rlabel_position(0)
    plt.yticks(ticks, [f"{int(round(t))}%" for t in ticks], color="grey", size=7)
    plt.ylim(0, axis_max)
    ax.set_rgrids(ticks, angle=0)

    n_profiles = len(all_weights_pct)
    line_alpha = 0.6 if n_profiles <= 5 else max(0.05, min(0.4, 20.0 / n_profiles))

    for w_pct in all_weights_pct:
        values = list(w_pct) + [w_pct[0]]
        ax.plot(angles, values, color=line_color, linewidth=line_width, alpha=line_alpha)

    scale_label = "autoscalato sul valore massimo osservato" if autoscale else "fisso 0–100%"
    ax.set_title(f"Profili composizionali di {n_profiles} formulazione/i generata/e\n"
                 f"(scala assi: {scale_label})", fontsize=11, pad=20)

    fig.tight_layout()
    fig.savefig(radar_path, dpi=150)

    plt.show()
    plt.close(fig)


def build_radar_controls(origin_names: List[str], all_weights_pct: List[np.ndarray],
                          radar_path: str) -> widgets.VBox:
    radar_render_out = widgets.Output()

    autoscale_checkbox = widgets.Checkbox(
        value=False, description="Autoscaling sull'origine con % massima"
    )
    color_picker = widgets.ColorPicker(
        concise=False, description="Colore linee:", value="lightgray"
    )
    width_slider = widgets.FloatSlider(
        value=0.6, min=0.1, max=3.0, step=0.1, description="Spessore linee:",
        style={"description_width": "initial"}
    )
    update_button = widgets.Button(description="Aggiorna grafico radar", button_style="warning")

    def _redraw(_=None):
        with radar_render_out:
            clear_output(wait=True)
            render_radar_chart(
                origin_names, all_weights_pct, radar_path,
                line_color=color_picker.value,
                line_width=float(width_slider.value),
                autoscale=bool(autoscale_checkbox.value),
            )

    update_button.on_click(_redraw)

    controls_box = widgets.VBox([
        widgets.HTML("<b>Opzioni grafico radar</b> (premi 'Aggiorna' per applicarle; "
                     "il file scaricabile viene aggiornato di conseguenza):"),
        autoscale_checkbox,
        color_picker,
        width_slider,
        update_button,
        radar_render_out,
    ])

    return controls_box


# ============================================================
# Pulsanti di download manuale (evitano download multipli/duplicati)
# ============================================================

def make_download_buttons(file_specs: List[Tuple[str, str]]) -> widgets.VBox:
    rows = []
    for path, label in file_specs:
        btn = widgets.Button(description=f"⬇️ Scarica: {label}", button_style="")
        btn_out = widgets.Output()

        def _make_handler(p=path, o=btn_out):
            def _handler(_b):
                with o:
                    clear_output()
                    if IN_COLAB:
                        files.download(p)
                        print(f"Download avviato: {p}")
                    else:
                        print("Download automatico disponibile solo in Google Colab.")
            return _handler

        btn.on_click(_make_handler())
        rows.append(widgets.VBox([btn, btn_out]))

    return widgets.VBox(rows)


# ============================================================
# NUOVO — Gestione del Target Probabilistico
# (logica "Cella 4", inclusa qui perché richiamata automaticamente
# al termine delle Opzioni 3/4; disponibile anche come cella
# indipendente più sotto)
# ============================================================

def normalize_var_name_for_match(name: str) -> str:
    return str(name).strip()

def load_probabilistic_target_from_upload() -> pd.DataFrame:
    if not IN_COLAB:
        raise RuntimeError("Upload via browser disponibile solo in Google Colab.")
    print("📂 Carica il file .xlsx del Target probabilistico (es. matrice Monte Carlo, molte righe).")
    uploaded = files.upload()
    fname = list(uploaded.keys())[0]
    df = pd.read_excel(io.BytesIO(uploaded[fname]), sheet_name=0, header=HEADER_ROW_INDEX)
    print(f"✅ File caricato: {fname} — {df.shape[0]} righe, {df.shape[1]} colonne.")
    return df

def is_target_montecarlo_classic_format(df: pd.DataFrame) -> bool:
    if df is None or df.shape[0] == 0 or df.shape[1] < 4:
        return False
    cols = {str(c).strip().lower() for c in df.columns[1:]}
    return {"median", "p05", "p95"}.issubset(cols)

def build_target_probabilistico_stats(df: pd.DataFrame) -> Tuple[pd.DataFrame, str]:
    """Riproduce la logica di MMv3_11 (_is_montecarlo_target_format /
    _build_target_from_plain_matrix): se il file ha già colonne
    Variable/median/p05/p95[/sigma] le usa direttamente, altrimenti
    tratta il file come matrice grezza multi-riga (una osservazione
    Monte Carlo per riga) e calcola le statistiche per colonna."""
    if is_target_montecarlo_classic_format(df):
        out = df.copy()
        out.columns = [str(c).strip() for c in out.columns]
        rename_map = {}
        for c in out.columns:
            cl = c.lower()
            if cl in {"median", "p05", "p95", "sigma"}:
                rename_map[c] = cl
        out = out.rename(columns=rename_map)
        out = out.rename(columns={out.columns[0]: "Variable"})
        if "sigma" not in out.columns:
            out["sigma"] = np.nan
        return out[["Variable", "median", "p05", "p95", "sigma"]], \
               "Monte Carlo classico (Variable/median/p05/p95[/sigma])"

    numeric_df = df.apply(pd.to_numeric, errors="coerce")
    valid_cols_mask = numeric_df.notna().any(axis=0)
    numeric_df = numeric_df.loc[:, valid_cols_mask]
    if numeric_df.shape[1] == 0:
        raise ValueError("Il file Target probabilistico non contiene colonne numeriche utilizzabili.")

    var_names = [normalize_var_name_for_match(c) for c in numeric_df.columns]
    out = pd.DataFrame({
        "Variable": var_names,
        "median": numeric_df.median(axis=0, skipna=True).to_numpy(dtype=float),
        "p05": numeric_df.quantile(0.05, axis=0, interpolation="linear").to_numpy(dtype=float),
        "p95": numeric_df.quantile(0.95, axis=0, interpolation="linear").to_numpy(dtype=float),
        "sigma": numeric_df.std(axis=0, ddof=1, skipna=True).to_numpy(dtype=float),
    })
    return out, f"matrice grezza multi-riga ({numeric_df.shape[0]} osservazioni per variabile)"

def align_target_probabilistico_columns(generated_var_cols: List[str],
                                         target_stats_df: pd.DataFrame) -> Dict:
    """Allineamento per nome, con normalizzazione minima (solo strip degli
    spazi) per gestire compound con spazi finali/iniziali nel nome, come
    osservato nel file allegato."""
    gen_norm = {normalize_var_name_for_match(c): c for c in generated_var_cols}
    tgt_norm = {normalize_var_name_for_match(v): v for v in target_stats_df["Variable"]}

    common_keys = [k for k in gen_norm if k in tgt_norm]
    only_generated = [gen_norm[k] for k in gen_norm if k not in tgt_norm]
    only_target = [tgt_norm[k] for k in tgt_norm if k not in gen_norm]

    return {
        "n_common": len(common_keys),
        "n_only_generated": len(only_generated),
        "n_only_target": len(only_target),
        "only_generated": only_generated,
        "only_target": only_target,
        "common_keys": common_keys,
    }

def build_target_probabilistico_section() -> widgets.VBox:
    """Sezione 'Gestione del Target Probabilistico': per ora si limita al
    caricamento del file, al riconoscimento del formato, al calcolo delle
    statistiche (mediana/p05/p95/sigma) e all'allineamento delle colonne
    con le variabili già generate in questa sessione."""

    title = widgets.HTML("<h3>Gestione del Target Probabilistico</h3>")
    load_prob_button = widgets.Button(description="Carica Target probabilistico (.xlsx)",
                                       button_style="info")
    prob_out = widgets.Output()

    def on_load_prob_clicked(_):
        with prob_out:
            clear_output()
            try:
                raw_df = load_probabilistic_target_from_upload()
            except Exception as e:
                print(f"❌ {e}")
                return

            target_stats_df, format_label = build_target_probabilistico_stats(raw_df)
            print(f"ℹ️ Formato riconosciuto: {format_label}")
            print(f"✅ Statistiche calcolate per {target_stats_df.shape[0]} variabili "
                  f"(mediana, p05, p95, sigma).")

            var_cols = state.get("var_cols", [])
            report = None
            if var_cols:
                report = align_target_probabilistico_columns(var_cols, target_stats_df)
                print("\n--- Allineamento colonne (compound) tra dataset generato e Target probabilistico ---")
                print(f"  Solo nel dataset generato (Cella 1): {report['n_only_generated']}")
                if report["only_generated"]:
                    print(f"    {report['only_generated']}")
                print(f"  Solo nel Target probabilistico caricato: {report['n_only_target']}")
                if report["only_target"]:
                    print(f"    {report['only_target']}")
                print(f"  Variabili comuni: {report['n_common']}")
            else:
                print("⚠️ Nessuna variabile generata trovata in 'state': allineamento non eseguibile ora.")

            print("\n--- Anteprima statistiche Target probabilistico (prime 10 variabili) ---")
            display(target_stats_df.head(10).round(4))

            state["target_probabilistico"] = {
                "stats_df": target_stats_df,
                "format_label": format_label,
                "alignment_report": report,
            }
            print("\n✅ Target probabilistico salvato in memoria (state['target_probabilistico']). "
                  "Le funzionalità di confronto/match con questo Target saranno aggiunte in un "
                  "prossimo passaggio.")

    load_prob_button.on_click(on_load_prob_clicked)
    return widgets.VBox([title, load_prob_button, prob_out])


# ============================================================
# UI — Step 1: titolo, scenario, tasto di caricamento file
# ============================================================

app_title = widgets.HTML("<h2>Validazione Miscele ricostruite con Algoritmo Genetico</h2>")

option_dropdown = widgets.Dropdown(
    options=[
        ("Opzione 1 — Target unitario, da Mono-Origine unitari", "opt1"),
        ("Opzione 2 — Target unitario, da Mono-Origine multi-batch", "opt2"),
        ("Opzione 3 — Target probabilistico, da Mono-Origine unitari", "opt3"),
        ("Opzione 4 — Target probabilistico, da Mono-Origine multi-batch", "opt4"),
    ],
    value="opt1",
    description="Scenario:",
    style={"description_width": "initial"},
    layout=widgets.Layout(width="520px")
)

load_button = widgets.Button(description="1) Carica file sorgente", button_style="info")
load_out = widgets.Output()
settings_box_container = widgets.VBox([])

display(widgets.VBox([app_title, option_dropdown, load_button, load_out, settings_box_container]))

state: Dict = {}


def on_load_clicked(_):
    with load_out:
        clear_output()
        settings_box_container.children = []
        opt = option_dropdown.value

        if opt in ("opt1", "opt3"):
            mono_df = load_xlsx_from_upload(
                "📂 Carica il file .xlsx con la matrice Mono-Origine (1 sola riga per origine)."
            )
            origin_col = detect_origin_column(mono_df)
            origin_names = mono_df[origin_col].astype(str).tolist()
            var_cols = list(mono_df.columns[START_COL_INDEX:])

            if mono_df.shape[0] != len(set(origin_names)):
                print("⚠️ Attenzione: alcune origini compaiono più di una volta. "
                      "Questa opzione richiede 1 sola riga per origine.")

            state.update(dict(option=opt, mono_df=mono_df, origin_col=origin_col,
                               origin_names=origin_names, var_cols=var_cols))

        elif opt in ("opt2", "opt4"):
            multi_df = load_xlsx_from_upload(
                "📂 Carica il file .xlsx Mono-Origine multi-batch (repliche per batch e origine)."
            )
            origin_col = detect_origin_column(multi_df)
            batch_col = detect_batch_column(multi_df, origin_col)
            var_cols = list(multi_df.columns[START_COL_INDEX:])

            n_before = multi_df.shape[0]
            excluded_mask = multi_df[origin_col].apply(is_excluded_origin)
            multi_df = multi_df.loc[~excluded_mask].reset_index(drop=False).rename(
                columns={"index": "__orig_index__"}
            )
            n_excluded = n_before - multi_df.shape[0]
            print(f"ℹ️ Righe escluse perché origine REF/QC: {n_excluded}")

            origin_names = sorted(multi_df[origin_col].astype(str).unique().tolist())
            print(f"✅ Colonna origine: '{origin_col}' | Colonna batch: '{batch_col}'")
            print(f"✅ Origini disponibili (escluse REF/QC): {len(origin_names)}")
            for name in origin_names:
                n_batches = multi_df.loc[multi_df[origin_col] == name, batch_col].nunique()
                n_rows = (multi_df[origin_col] == name).sum()
                print(f"   {name}: {n_batches} batch, {n_rows} righe/repliche totali")

            state.update(dict(option=opt, multi_df=multi_df, origin_col=origin_col,
                               batch_col=batch_col, origin_names=origin_names, var_cols=var_cols))

        else:
            print("⚠️ Questa opzione non è ancora implementata in questa versione.")
            return

        build_settings_form(state["origin_names"])


# ============================================================
# UI — Step 2: tabella Min/Max editabile + settaggi + tasto finale
# ============================================================

def build_settings_form(origin_names: List[str]):
    minmax_title = widgets.HTML(
        "<b>Limiti Min. % / Max. % per origine</b><br>"
        "Valori pre-impostati come da tabella di riferimento MixManager — modificabili."
    )

    header_row = widgets.HBox([
        widgets.Label("Origine", layout=widgets.Layout(width="220px")),
        widgets.Label("Min. %", layout=widgets.Layout(width="90px")),
        widgets.Label("Max. %", layout=widgets.Layout(width="90px")),
    ])

    min_widgets: Dict[str, widgets.BoundedFloatText] = {}
    max_widgets: Dict[str, widgets.BoundedFloatText] = {}
    minmax_rows = [header_row]

    for name in origin_names:
        default_min, default_max = get_default_minmax(name)
        min_w = widgets.BoundedFloatText(value=default_min, min=0.0, max=100.0, step=0.5,
                                          layout=widgets.Layout(width="90px"))
        max_w = widgets.BoundedFloatText(value=default_max, min=0.0, max=100.0, step=0.5,
                                          layout=widgets.Layout(width="90px"))
        min_widgets[name] = min_w
        max_widgets[name] = max_w
        minmax_rows.append(widgets.HBox([
            widgets.Label(name, layout=widgets.Layout(width="220px")), min_w, max_w
        ]))

    minmax_table = widgets.VBox(minmax_rows, layout=widgets.Layout(border="1px solid #ddd", padding="6px"))

    n_targets_widget = widgets.BoundedIntText(value=1, min=1, max=5000, step=1,
                                               description="N° Target unitari:",
                                               style={"description_width": "initial"})
    seed_widget = widgets.IntText(value=42, description="random_seed")

    threshold_note = widgets.HTML(
        f"<i>Con N° Target > {STACK_THRESHOLD}, l'output passa automaticamente al formato impilato "
        f"con colonna Target_ID, più istogrammi (bin 1%) e statistiche di distribuzione per origine, "
        f"mostrati qui sotto e scaricabili.<br>"
        f"Con N° Target > {VERIFICATION_THRESHOLD}, sarà inoltre possibile eseguire la cella successiva "
        f"'Verifica Algoritmo Genetico'.<br>"
        f"In ogni caso (anche con un solo Target) viene generato un grafico radar dei profili "
        f"composizionali per origine, con opzioni di personalizzazione dopo la prima visualizzazione.<br>"
        f"Con le Opzioni 3/4 (Target probabilistico), al termine della generazione si passa "
        f"automaticamente alla sezione 'Gestione del Target Probabilistico'.</i>"
    )

    run_button = widgets.Button(description="Genera Target unitari", button_style="success")
    run_out = widgets.Output()
    download_out = widgets.Output()

    settings_box_container.children = [
        widgets.VBox([
            widgets.HTML("<hr>"),
            minmax_title,
            minmax_table,
            widgets.HTML("<hr>"),
            n_targets_widget,
            seed_widget,
            threshold_note,
            run_button,
            run_out,
            download_out
        ])
    ]

    state["min_widgets"] = min_widgets
    state["max_widgets"] = max_widgets
    state["n_targets_widget"] = n_targets_widget
    state["seed_widget"] = seed_widget
    state["run_out"] = run_out
    state["download_out"] = download_out

    run_button.on_click(on_run_clicked)


# ============================================================
# Callback finale: generazione dei Target unitari (per tutte le opzioni)
# ============================================================

def on_run_clicked(_):
    run_out = state["run_out"]
    download_out = state["download_out"]

    with download_out:
        clear_output()

    with run_out:
        clear_output()

        opt = state["option"]
        origin_names = state["origin_names"]
        var_cols = state["var_cols"]
        min_widgets = state["min_widgets"]
        max_widgets = state["max_widgets"]
        n_targets = int(state["n_targets_widget"].value)
        rng = np.random.default_rng(int(state["seed_widget"].value))

        mins, maxs = [], []
        print("--- Limiti Min/Max % per origine (valori impostati dall'utente) ---")
        for name in origin_names:
            mn = float(min_widgets[name].value)
            mx = float(max_widgets[name].value)
            if mn > mx:
                mn, mx = mx, mn
            mins.append(mn)
            maxs.append(mx)
            print(f"  {name}: {mn:.1f}–{mx:.1f}%")
        mins = np.array(mins, dtype=float)
        maxs = np.array(maxs, dtype=float)

        sheets_file1: Dict[str, pd.DataFrame] = {}
        sheets_file2: Dict[str, pd.DataFrame] = {}
        all_weights_pct: List[np.ndarray] = []
        origin_field_name = None

        for i in range(n_targets):
            label = target_letter(i)
            sheet_name = f"Target unitario {label}"

            try:
                w_true_pct = sample_bounded_composition(mins, maxs, total=100.0, rng=rng)
            except ValueError as e:
                print(f"❌ {e}")
                print("   Modifica i limiti Min/Max nella tabella e riprova.")
                return

            all_weights_pct.append(w_true_pct.copy())

            if opt in ("opt1", "opt3"):
                sheet1, sheet2, origin_field_name = build_target_sheets_opt1(
                    state["mono_df"], var_cols, w_true_pct
                )
            elif opt in ("opt2", "opt4"):
                sheet1, sheet2, origin_field_name = build_target_sheets_opt2(
                    state["multi_df"], state["origin_col"], state["batch_col"],
                    var_cols, origin_names, w_true_pct, rng
                )
            else:
                print("⚠️ Opzione non gestita.")
                return

            sheets_file1[sheet_name] = sheet1
            sheets_file2[sheet_name] = sheet2

            if n_targets <= STACK_THRESHOLD:
                print(f"\n=== Target unitario {label} ({i+1}/{n_targets}) ===")
                for name, w in zip(origin_names, w_true_pct):
                    print(f"  {name}: {w:.2f}%")
                print(f"  Somma totale: {w_true_pct.sum():.4f}%")
            elif (i + 1) % 10 == 0 or (i + 1) == n_targets:
                print(f"Generati {i + 1}/{n_targets} Target...")

        out_path_1 = "MonoOrigine_con_Perc_Target.xlsx"
        out_path_2 = "Target_unitario.xlsx"
        download_specs: List[Tuple[str, str]] = []

        # ------------------------------------------------------------
        # MODALITÀ A) N <= soglia: un foglio per Target
        # ------------------------------------------------------------
        if n_targets <= STACK_THRESHOLD:
            with pd.ExcelWriter(out_path_1) as writer:
                for sheet_name, df_sheet in sheets_file1.items():
                    df_sheet.to_excel(writer, sheet_name=sheet_name[:31], index=False)
            with pd.ExcelWriter(out_path_2) as writer:
                for sheet_name, df_sheet in sheets_file2.items():
                    df_sheet.to_excel(writer, sheet_name=sheet_name[:31], index=False)

            print(f"\n✅ Creato {out_path_1} — {n_targets} fogli.")
            print(f"✅ Creato {out_path_2} — {n_targets} fogli.")

            download_specs.append((out_path_1, "matrice Mono-Origine + Perc_Target_%"))
            download_specs.append((out_path_2, "Target unitario"))

        # ------------------------------------------------------------
        # MODALITÀ B) N > soglia: righe impilate, un solo foglio per file
        # ------------------------------------------------------------
        else:
            print(f"\nℹ️ N° Target ({n_targets}) > {STACK_THRESHOLD}: passo al formato impilato.")

            file2_rows = []
            for i, (sheet_name, df_sheet) in enumerate(sheets_file2.items()):
                label = target_letter(i)
                row = df_sheet.iloc[0].to_dict()
                row = {"Target_ID": f"Target_unit_{label}", **row}
                file2_rows.append(row)
            stacked_file2 = pd.DataFrame(file2_rows)
            stacked_file2.to_excel(out_path_2, index=False)

            block_frames = []
            for i, (sheet_name, df_sheet) in enumerate(sheets_file1.items()):
                label = target_letter(i)
                block = df_sheet.copy()
                block.insert(0, "Target_ID", f"Target_unit_{label}")
                block_frames.append(block)
                if i < len(sheets_file1) - 1:
                    blank_row = pd.DataFrame([{c: np.nan for c in block.columns}])
                    block_frames.append(blank_row)
            stacked_file1 = pd.concat(block_frames, ignore_index=True)
            stacked_file1.to_excel(out_path_1, index=False)

            print(f"✅ Creato {out_path_1} — formato impilato, {n_targets} blocchi separati da riga vuota.")
            print(f"✅ Creato {out_path_2} — formato impilato, {n_targets} righe (una per Target).")

            state["stacked_file1"] = stacked_file1
            state["stacked_file2"] = stacked_file2
            state["origin_field_name"] = origin_field_name
            state["var_cols"] = var_cols
            state["origin_names"] = origin_names
            state["n_targets"] = n_targets

            if n_targets > VERIFICATION_THRESHOLD:
                print(f"\n✅ N° Target ({n_targets}) > {VERIFICATION_THRESHOLD}: puoi ora eseguire la cella "
                      f"successiva 'Verifica Algoritmo Genetico'.")

            pct_by_origin: Dict[str, List[float]] = {name: [] for name in origin_names}
            for df_sheet in sheets_file1.values():
                for _, row in df_sheet.iterrows():
                    pct_by_origin[str(row[origin_field_name])].append(float(row["Perc_Target_%"]))

            hist_zip_path = "istogrammi_distribuzione_target.zip"
            stats_xlsx_path = "statistiche_distribuzione_target.xlsx"
            build_and_show_distributions(pct_by_origin, hist_zip_path, stats_xlsx_path)
            print(f"✅ Creato {stats_xlsx_path} e {hist_zip_path}.")

            download_specs.append((out_path_1, "matrice Mono-Origine + Perc_Target_% (impilato)"))
            download_specs.append((out_path_2, "Target unitario (impilato)"))
            download_specs.append((stats_xlsx_path, "statistiche di distribuzione per origine"))
            download_specs.append((hist_zip_path, "istogrammi per origine (zip)"))

        # ------------------------------------------------------------
        # Radar/spider chart — sempre generato, indipendentemente da N
        # ------------------------------------------------------------
        radar_path = "radar_profili_composizionali.png"
        print("\n--- Radar chart: profilo composizionale per formulazione ---")
        render_radar_chart(origin_names, all_weights_pct, radar_path)
        download_specs.append((radar_path, "radar chart profili composizionali"))

        radar_controls = build_radar_controls(origin_names, all_weights_pct, radar_path)
        display(radar_controls)

        print("\n⚠️ Ricorda: i file Target_unitario NON contengono i pesi reali — vanno confrontati")
        print("   solo a posteriori con Perc_Target_% del file Mono-Origine, dopo la ricostruzione via GA.")

        with download_out:
            display(widgets.HTML("<hr><b>Download file generati:</b>"))
            display(make_download_buttons(download_specs))

        if opt in ("opt3", "opt4"):
            print("\n" + "=" * 70)
            print("Opzione con Target probabilistico: generazione completata.")
            print("👉 Esegui ora la Cella 4 — 'Gestione del Target Probabilistico' — "
                  "per impostare i limiti Min/Max dedicati e caricare il Target.")
            print("=" * 70)

load_button.on_click(on_load_clicked)
```

```python
# @title
# ============================================================
# Validazione Miscele ricostruite con Algoritmo Genetico
# ------------------------------------------------------------
# CELLA 3 — MATCH_% per Target singolo (non probabilistico)
#
# Adatta il criterio MATCH_% originale di MixManager (pensato per
# target Monte Carlo) al caso di Target puntuale, e lo usa come
# QUARTO criterio (accanto a distanza euclidea, media delta
# assoluto, coseno) per scegliere quale delle N formulazioni
# generate in Cella 1 meglio approssima il Target_unitario_reale
# caricato in Cella 2.
#
# NUOVO: tabella di confronto composizione % per origine, una
# colonna per criterio (4 colonne), con la formulazione vincente
# secondo ciascun criterio — scaricabile.
#
# Richiede l'esecuzione di Cella 1 e Cella 2 (con relativo patch
# a state["verification_results"]) nella stessa sessione.
# ============================================================

import numpy as np
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output

try:
    from google.colab import files
    IN_COLAB = True
except Exception:
    IN_COLAB = False

EPS = 1e-9

# ---------- Default storici (stessi della formula originale MixManager) ----------
K_DEFAULT = 2.0
Z_CAP_DEFAULT = 4.0


# ---------- Verifica prerequisiti da Cella 1 + Cella 2 ----------
_required_keys_cella1 = ["stacked_file1", "origin_field_name", "origin_names"]
_required_keys_cella2 = ["verification_results"]
_missing_c1 = [k for k in _required_keys_cella1 if k not in state]
_missing_c2 = [k for k in _required_keys_cella2 if k not in state]
_is_probabilistic = "state" in globals() and state.get("option") in ("opt3", "opt4")

_vr = state.get("verification_results", {})
_required_keys_vr = ["real_target", "profiles_common", "target_ids", "scale_distanza",
                      "common_var_cols", "winner_id", "metrics_df"]
_missing_vr = [k for k in _required_keys_vr if k not in _vr]

cella3_gate_out = widgets.Output()
display(cella3_gate_out)

with cella3_gate_out:
    if _is_probabilistic:
        print("↪️ Opzione 3/4 (Target probabilistico) selezionata in Cella 1: questa cella "
              "('MATCH_% per Target singolo') si applica solo al Target puntuale (Opzioni 1/2) "
              "e viene bypassata. Procedi con la Cella 4 — 'Gestione del Target Probabilistico'.")
    elif _missing_c1 or _missing_c2:
        print("❌ Esegui prima la Cella 1 e la Cella 2 (in questo ordine), nella stessa sessione. "
              f"Chiavi mancanti: {_missing_c1 + _missing_c2}")
    elif _missing_vr:
        print("❌ La Cella 2 è stata eseguita, ma manca il patch che salva i dati necessari "
              "('real_target', 'profiles_common', 'target_ids', 'scale_distanza') in "
              "state['verification_results']. Applica il patch indicato, rilancia la Cella 2 "
              "(ricaricando di nuovo il Target_unitario_reale), poi rilancia questa cella.")
    else:
        print(f"✅ Trovate {state['n_targets']} formulazioni generate nella Cella 1. "
              "Procedi caricando il Target_unitario_reale qui sotto.")


# ============================================================
# Formula MATCH_% adattata al Target singolo (non probabilistico)
# ============================================================

def compute_matching_percent_puntuale(mix_profile: np.ndarray, target_profile: np.ndarray,
                                       scale: np.ndarray, K: float, Z_CAP: float) -> float:
    """Versione depurata di compute_matching_percent (MixManager originale):
    - RIMOSSA la penalità 'fuori banda P05-P95' (con Target puntuale la banda
      ha ampiezza zero e duplicherebbe la distanza centrale).
    - target_sigma sostituita da `scale`: la dispersione REALE osservata tra
      le formulazioni generate (stessa scala già usata per la distanza euclidea
      in Cella 2), al posto del fallback fisso 1.0 uguale per ogni variabile.
    """
    sigma_for_z = np.where(scale < EPS, 1.0, scale)
    z = (mix_profile - target_profile) / sigma_for_z
    z = np.clip(z, -Z_CAP, Z_CAP)
    dist = float(np.sqrt(np.mean(z ** 2)))

    scale_k = max(K, EPS)
    score = 100.0 / (1.0 + dist / scale_k)
    return float(np.clip(score, 0.0, 100.0))


def compute_match_percent_ranking(profiles: np.ndarray, target_ids: list,
                                   real_target: np.ndarray, scale: np.ndarray,
                                   K: float, Z_CAP: float) -> pd.DataFrame:
    """Calcola lo score interno per ciascuna delle N formulazioni, poi calibra
    il MATCH_% usando la MEDIANA degli score interni delle N formulazioni stesse
    come ancora (MATCH_%=0), esattamente come faceva calibrate_match_percent
    nell'originale — ma qui la baseline random è la popolazione già generata
    dalla Cella 1, senza bisogno di generarne una separata."""

    internal_scores = np.array([
        compute_matching_percent_puntuale(p, real_target, scale, K, Z_CAP)
        for p in profiles
    ])

    baseline_median_internal = float(np.median(internal_scores))
    denom = max(100.0 - baseline_median_internal, 1e-6)

    def calibrate(score_internal: float) -> float:
        if score_internal <= baseline_median_internal:
            return 0.0
        val = (score_internal - baseline_median_internal) * 100.0 / denom
        return float(np.clip(val, 0.0, 100.0))

    match_pct = np.array([calibrate(s) for s in internal_scores])

    return pd.DataFrame({
        "Target_ID": target_ids,
        "Score_interno": internal_scores,
        "MATCH_%": match_pct,
    })


# ============================================================
# Composizione % per origine di una data formulazione
# ============================================================

def get_composition_by_origin(stacked_file1: pd.DataFrame, origin_field_name: str,
                               target_id: str, origin_names: list) -> dict:
    block = stacked_file1[stacked_file1["Target_ID"] == target_id]
    pct_by_origin = dict(zip(block[origin_field_name].astype(str), block["Perc_Target_%"].astype(float)))
    return {name: pct_by_origin.get(name, np.nan) for name in origin_names}


# ============================================================
# UI — parametri K e Z_CAP, con spiegazione, e calcolo
# ============================================================

if not _is_probabilistic and not _missing_c1 and not _missing_c2 and not _missing_vr:

    cella3_title = widgets.HTML("<h3>MATCH_% per Target singolo — quarto criterio di confronto</h3>")

    params_explanation = widgets.HTML(f"""
    <div style="border:1px solid #ddd; padding:10px; background:#fafafa;">
    <b>Cosa fanno i due parametri, in parole semplici:</b>
    <ul>
      <li><b>K (scala di sensibilità, default {K_DEFAULT})</b> — controlla quanto velocemente il punteggio
      scende all'aumentare della distanza dal Target (score = 100 / (1 + distanza/K)).<br>
      &nbsp;&nbsp;• <b>K più alto</b> → il punteggio scende più lentamente: sei più "permissivo",
      anche formulazioni non perfette ottengono un punteggio ragionevole.<br>
      &nbsp;&nbsp;• <b>K più basso</b> → il punteggio scende rapidamente: sei più "severo",
      solo le formulazioni molto vicine al Target ottengono punteggi alti.</li>
      <li><b>Z_CAP (limite per variabile, default {Z_CAP_DEFAULT})</b> — prima di calcolare la distanza,
      lo scostamento di ciascuna variabile chimica viene "tagliato" a questo valore massimo
      (espresso in deviazioni standard rispetto alla dispersione osservata tra le formulazioni).<br>
      &nbsp;&nbsp;• <b>Z_CAP più alto</b> → una singola variabile molto fuori scala può pesare di più
      e dominare il punteggio complessivo (meno protezione dagli outlier).<br>
      &nbsp;&nbsp;• <b>Z_CAP più basso</b> → limiti quanto una singola variabile "impazzita" può
      influenzare il punteggio totale, ma rischi di mascherare uno scostamento realmente importante
      su quella variabile.</li>
    </ul>
    I valori di default coincidono con quelli storicamente usati nel workflow probabilistico originale.
    </div>
    """)

    k_slider = widgets.FloatSlider(value=K_DEFAULT, min=0.1, max=10.0, step=0.1,
                                    description="K:", style={"description_width": "initial"})
    zcap_slider = widgets.FloatSlider(value=Z_CAP_DEFAULT, min=0.5, max=10.0, step=0.5,
                                       description="Z_CAP:", style={"description_width": "initial"})

    calc_button = widgets.Button(description="Calcola MATCH_% (Target singolo)", button_style="success")
    calc_out = widgets.Output()
    calc_download_out = widgets.Output()

    display(widgets.VBox([
        cella3_title, params_explanation, k_slider, zcap_slider, calc_button, calc_out, calc_download_out
    ]))

    def on_calc_clicked(_):
        with calc_download_out:
            clear_output()

        with calc_out:
            clear_output()

            vr = state["verification_results"]
            real_target = vr["real_target"]
            profiles = vr["profiles_common"]
            target_ids = vr["target_ids"]
            scale = vr["scale_distanza"]
            common_var_cols = vr["common_var_cols"]
            official_winner_id = vr["winner_id"]
            metrics_df = vr["metrics_df"]

            stacked_file1 = state["stacked_file1"]
            origin_field_name = state["origin_field_name"]
            origin_names = state["origin_names"]

            K = float(k_slider.value)
            Z_CAP = float(zcap_slider.value)

            print(f"--- Parametri usati: K = {K:.2f} | Z_CAP = {Z_CAP:.2f} ---")

            ranking_df = compute_match_percent_ranking(profiles, target_ids, real_target, scale, K, Z_CAP)
            ranking_df_sorted = ranking_df.sort_values("MATCH_%", ascending=False).reset_index(drop=True)

            match_winner_id = ranking_df_sorted.iloc[0]["Target_ID"]

            print(f"\n🏆 Formulazione con MATCH_% più alto: {match_winner_id} "
                  f"(MATCH_% = {ranking_df_sorted.iloc[0]['MATCH_%']:.2f}, "
                  f"score interno = {ranking_df_sorted.iloc[0]['Score_interno']:.2f})")

            print("\n--- Top 5 per MATCH_% (Target singolo) ---")
            display(ranking_df_sorted.head(5).round(3))

            if match_winner_id != official_winner_id:
                print(f"\n⚠️ Attenzione: il vincitore secondo MATCH_% ({match_winner_id}) non coincide "
                      f"con la vincitrice ufficiale per distanza euclidea normalizzata "
                      f"({official_winner_id}, calcolata in Cella 2). "
                      "Le due metriche pesano gli scostamenti in modo diverso (in particolare, MATCH_% "
                      "applica il taglio Z_CAP e una trasformazione non lineare della distanza) — "
                      "valuta entrambe prima di scegliere quale formulazione adottare.")
            else:
                print(f"\n✅ Il vincitore per MATCH_% coincide con la vincitrice ufficiale ({official_winner_id}).")

            # --- Composizione % della formulazione vincente per MATCH_% ---
            known_meta_cols = [c for c in ["Target_ID", origin_field_name, "Batch_selezionato",
                                            "Indice_riga_originale", "N_repliche_disponibili_nel_batch",
                                            "Perc_Target_%"] if c in stacked_file1.columns]
            match_winner_block = stacked_file1[stacked_file1["Target_ID"] == match_winner_id].copy()
            match_winner_composition = match_winner_block[known_meta_cols].reset_index(drop=True)

            print(f"\n--- Composizione % della formulazione vincente per MATCH_% ({match_winner_id}) ---")
            display(match_winner_composition.round(4))

            # ============================================================
            # Tabella di confronto: 1 riga per origine, 1 colonna per
            # criterio, cella = % di impiego secondo la formulazione
            # vincente per quel criterio
            # ============================================================

            winner_eucl_id = official_winner_id
            winner_mad_id = metrics_df.sort_values(
                "Media_delta_assoluto_norm", ascending=True
            ).iloc[0]["Target_ID"]
            winner_cos_id = metrics_df.sort_values(
                "Cosine_similarity", ascending=False
            ).iloc[0]["Target_ID"]
            winner_match_id = match_winner_id

            criteri_winner = {
                f"Distanza_euclidea_norm ({winner_eucl_id})": winner_eucl_id,
                f"Media_delta_assoluto_norm ({winner_mad_id})": winner_mad_id,
                f"Cosine_similarity ({winner_cos_id})": winner_cos_id,
                f"MATCH_% ({winner_match_id})": winner_match_id,
            }

            comparison_table = pd.DataFrame(index=origin_names)
            for col_label, tid in criteri_winner.items():
                comp = get_composition_by_origin(stacked_file1, origin_field_name, tid, origin_names)
                comparison_table[col_label] = [comp[name] for name in origin_names]

            comparison_table.index.name = "Origine"
            comparison_table_reset = comparison_table.reset_index()

            print("\n--- Tabella di confronto: composizione % per origine, una colonna per criterio ---")
            display(comparison_table_reset.round(3))

            totals_check = comparison_table.sum(axis=0)
            print("\nControllo somme (dovrebbero essere ≈100 per ciascuna colonna):")
            for col_label, tot in totals_check.items():
                print(f"  {col_label}: {tot:.4f}%")

            # --- Salvataggio output (xlsx con tutti i risultati della cella) ---
            results_xlsx_path = "MATCH_percent_target_singolo.xlsx"
            params_df = pd.DataFrame({"Parametro": ["K", "Z_CAP"], "Valore": [K, Z_CAP]})

            with pd.ExcelWriter(results_xlsx_path) as writer:
                params_df.to_excel(writer, sheet_name="Parametri_usati", index=False)
                ranking_df_sorted.to_excel(writer, sheet_name="Ranking_MATCH_percent", index=False)
                match_winner_composition.to_excel(writer, sheet_name="Composizione_vincente_MATCH", index=False)
                comparison_table_reset.to_excel(writer, sheet_name="Confronto_4_criteri", index=False)

            print(f"\n✅ Creato {results_xlsx_path} (Parametri_usati, Ranking_MATCH_percent, "
                  f"Composizione_vincente_MATCH, Confronto_4_criteri).")

            with calc_download_out:
                display(widgets.HTML("<hr><b>Download risultati:</b>"))

                btn1 = widgets.Button(description="⬇️ Scarica: risultati MATCH_% Target singolo")
                btn1_out = widgets.Output()

                def _dl1(_b, p=results_xlsx_path, o=btn1_out):
                    with o:
                        clear_output()
                        if IN_COLAB:
                            files.download(p)
                            print(f"Download avviato: {p}")
                        else:
                            print("Download automatico disponibile solo in Google Colab.")

                btn1.on_click(_dl1)

                # --- pulsante dedicato per la sola tabella di confronto 4 criteri ---
                comparison_xlsx_path = "Confronto_4_criteri_composizione.xlsx"
                comparison_table_reset.to_excel(comparison_xlsx_path, index=False)

                btn2 = widgets.Button(description="⬇️ Scarica: tabella confronto 4 criteri (composizione %)")
                btn2_out = widgets.Output()

                def _dl2(_b, p=comparison_xlsx_path, o=btn2_out):
                    with o:
                        clear_output()
                        if IN_COLAB:
                            files.download(p)
                            print(f"Download avviato: {p}")
                        else:
                            print("Download automatico disponibile solo in Google Colab.")

                btn2.on_click(_dl2)

                display(widgets.VBox([btn1, btn1_out, btn2, btn2_out]))

            state["match_percent_results"] = {
                "ranking_df": ranking_df_sorted,
                "match_winner_id": match_winner_id,
                "K": K,
                "Z_CAP": Z_CAP,
                "comparison_table": comparison_table_reset,
            }

    calc_button.on_click(on_calc_clicked)
```

```python
# @title
# ============================================================
# Validazione Miscele ricostruite con Algoritmo Genetico
# ------------------------------------------------------------
# CELLA 4 — Gestione del Target Probabilistico
#
# Cella indipendente. Richiede che la Cella 1 sia già stata eseguita
# (Opzione 3 o 4), nella stessa sessione, per avere disponibili in
# `state`: origin_names, var_cols; e in sessione: get_default_minmax,
# normalize_origin_key (definite in Cella 1).
#
# Funzionalità:
#   1) Tabella Min/Max % per origine — stessa della Cella 1, stessi
#      default, editabile
#   2) Caricamento del file Target probabilistico (.xlsx): riconosce
#      il formato (Monte Carlo classico Variable/median/p05/p95[/sigma]
#      oppure matrice grezza multi-riga), calcola mediana/p05/p95/sigma
#      per variabile
#   3) Allineamento colonne con le variabili generate in Cella 1
#   4) Metriche di densità di probabilità in funzione della distanza
#      dal centroide (mediana), calcolabili solo se disponibile la
#      matrice grezza multi-riga
#   5) NUOVO: istogramma distanza-dal-centroide sempre mostrato subito,
#      con opzione di cut-off percentile (default P95, modificabile) per
#      escludere le osservazioni più marginali, e pulsante di refresh
#      dedicato che ridisegna il grafico e aggiorna il file scaricabile
#   6) Tutto mostrato inline + scaricabile con pulsanti manuali
# ============================================================

import io
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis
from typing import Dict, List, Tuple, Optional
import ipywidgets as widgets
from IPython.display import display, clear_output

try:
    from google.colab import files
    IN_COLAB = True
except Exception:
    IN_COLAB = False

EPS = 1e-9
HEADER_ROW_INDEX = 0
CUTOFF_PERCENTILE_DEFAULT = 95.0

# ---------- Verifica prerequisiti dalla Cella 1 ----------
_gate_out = widgets.Output()
display(_gate_out)

_ready = True
with _gate_out:
    if "state" not in globals():
        print("❌ Esegui prima la Cella 1 (Opzione 3 o 4), nella stessa sessione.")
        _ready = False
    elif "origin_names" not in state or "var_cols" not in state:
        print("❌ Dati mancanti da 'state': esegui prima la Cella 1 (Opzione 3 o 4) "
              "fino al termine della generazione.")
        _ready = False
    elif "get_default_minmax" not in globals():
        print("❌ Funzione 'get_default_minmax' non trovata: esegui prima la Cella 1 "
              "nella stessa sessione (non solo in una sessione precedente).")
        _ready = False
    else:
        print("✅ Dati disponibili dalla Cella 1. Procedi qui sotto.")


# ============================================================
# Funzioni di supporto (autonome)
# ============================================================

def load_probabilistic_target_from_upload() -> pd.DataFrame:
    if not IN_COLAB:
        raise RuntimeError("Upload via browser disponibile solo in Google Colab.")
    print("📂 Carica il file .xlsx del Target probabilistico (es. matrice Monte Carlo, molte righe).")
    uploaded = files.upload()
    fname = list(uploaded.keys())[0]
    df = pd.read_excel(io.BytesIO(uploaded[fname]), sheet_name=0, header=HEADER_ROW_INDEX)
    print(f"✅ File caricato: {fname} — {df.shape[0]} righe, {df.shape[1]} colonne.")
    return df


def normalize_var_name_for_match(name: str) -> str:
    return str(name).strip()


def is_target_montecarlo_classic_format(df: pd.DataFrame) -> bool:
    if df is None or df.shape[0] == 0 or df.shape[1] < 4:
        return False
    cols = {str(c).strip().lower() for c in df.columns[1:]}
    return {"median", "p05", "p95"}.issubset(cols)


def build_target_probabilistico_stats(df: pd.DataFrame) -> Tuple[pd.DataFrame, str]:
    """Stessa logica di MMv3_11 (_is_montecarlo_target_format /
    _build_target_from_plain_matrix): se il file ha già colonne
    Variable/median/p05/p95[/sigma] le usa direttamente, altrimenti
    tratta il file come matrice grezza multi-riga (una osservazione
    Monte Carlo per riga) e calcola le statistiche per colonna."""
    if is_target_montecarlo_classic_format(df):
        out = df.copy()
        out.columns = [str(c).strip() for c in out.columns]
        rename_map = {}
        for c in out.columns:
            cl = c.lower()
            if cl in {"median", "p05", "p95", "sigma"}:
                rename_map[c] = cl
        out = out.rename(columns=rename_map)
        out = out.rename(columns={out.columns[0]: "Variable"})
        if "sigma" not in out.columns:
            out["sigma"] = np.nan
        return out[["Variable", "median", "p05", "p95", "sigma"]], \
               "Monte Carlo classico (Variable/median/p05/p95[/sigma])"

    numeric_df = df.apply(pd.to_numeric, errors="coerce")
    valid_cols_mask = numeric_df.notna().any(axis=0)
    numeric_df = numeric_df.loc[:, valid_cols_mask]
    if numeric_df.shape[1] == 0:
        raise ValueError("Il file Target probabilistico non contiene colonne numeriche utilizzabili.")

    var_names = [normalize_var_name_for_match(c) for c in numeric_df.columns]
    out = pd.DataFrame({
        "Variable": var_names,
        "median": numeric_df.median(axis=0, skipna=True).to_numpy(dtype=float),
        "p05": numeric_df.quantile(0.05, axis=0, interpolation="linear").to_numpy(dtype=float),
        "p95": numeric_df.quantile(0.95, axis=0, interpolation="linear").to_numpy(dtype=float),
        "sigma": numeric_df.std(axis=0, ddof=1, skipna=True).to_numpy(dtype=float),
    })
    return out, f"matrice grezza multi-riga ({numeric_df.shape[0]} osservazioni per variabile)"


def align_target_probabilistico_columns(generated_var_cols: List[str],
                                         target_stats_df: pd.DataFrame) -> Dict:
    gen_norm = {normalize_var_name_for_match(c): c for c in generated_var_cols}
    tgt_norm = {normalize_var_name_for_match(v): v for v in target_stats_df["Variable"]}

    common_keys = [k for k in gen_norm if k in tgt_norm]
    only_generated = [gen_norm[k] for k in gen_norm if k not in tgt_norm]
    only_target = [tgt_norm[k] for k in tgt_norm if k not in gen_norm]

    return {
        "n_common": len(common_keys),
        "n_only_generated": len(only_generated),
        "n_only_target": len(only_target),
        "only_generated": only_generated,
        "only_target": only_target,
        "common_keys": common_keys,
    }


def compute_centroid_distance_stats(distances: np.ndarray) -> Dict:
    """Statistiche descrittive sul vettore di distanze fornito (utile sia
    per il dataset completo sia per quello dopo cut-off), pensate per
    essere riutilizzate in futuro per modellare la densità di probabilità
    in funzione della distanza dal centroide."""
    return {
        "N_osservazioni": int(len(distances)),
        "Distanza_media": float(np.mean(distances)),
        "Distanza_mediana": float(np.median(distances)),
        "Distanza_dev_std": float(np.std(distances, ddof=1)) if len(distances) > 1 else 0.0,
        "Distanza_min": float(np.min(distances)),
        "Distanza_P05": float(np.percentile(distances, 5)),
        "Distanza_P25": float(np.percentile(distances, 25)),
        "Distanza_P75": float(np.percentile(distances, 75)),
        "Distanza_P95": float(np.percentile(distances, 95)),
        "Distanza_P99": float(np.percentile(distances, 99)),
        "Distanza_max": float(np.max(distances)),
        "Distanza_skewness": float(skew(distances)) if len(distances) > 2 else np.nan,
        "Distanza_curtosi": float(kurtosis(distances)) if len(distances) > 3 else np.nan,
    }


def compute_raw_distances(numeric_df: pd.DataFrame, target_stats_df: pd.DataFrame) -> np.ndarray:
    """Distanza normalizzata (RMS degli z-score per variabile, sigma del
    Target come scala) di ciascuna osservazione Monte Carlo dal centroide
    (mediana) del Target."""
    median_vec = target_stats_df["median"].to_numpy(dtype=float)
    sigma_vec = target_stats_df["sigma"].to_numpy(dtype=float)
    sigma_safe = np.where(np.isfinite(sigma_vec) & (sigma_vec > EPS), sigma_vec, 1.0)

    X = numeric_df.to_numpy(dtype=float)
    Z = (X - median_vec.reshape(1, -1)) / sigma_safe.reshape(1, -1)
    return np.sqrt(np.mean(Z ** 2, axis=1))


def render_centroid_distance_histogram(distances_full: np.ndarray, cutoff_percentile: float,
                                        chart_path: str) -> Tuple[np.ndarray, float]:
    """Disegna l'istogramma delle distanze, evidenziando con una linea
    verticale la soglia di cut-off al percentile scelto. Ritorna il
    sottoinsieme di distanze entro il cut-off + il valore di soglia."""
    threshold = float(np.percentile(distances_full, cutoff_percentile))
    kept_mask = distances_full <= threshold
    distances_kept = distances_full[kept_mask]
    n_excluded = int((~kept_mask).sum())

    fig, ax = plt.subplots(figsize=(7, 4.5))
    ax.hist(distances_full, bins=50, edgecolor="black", alpha=0.55, label="Tutte le osservazioni")
    ax.hist(distances_kept, bins=50, edgecolor="black", alpha=0.75, color="tab:blue",
             label=f"Entro P{cutoff_percentile:g} (mantenute)")
    ax.axvline(threshold, color="red", linewidth=2,
               label=f"Cut-off P{cutoff_percentile:g} = {threshold:.3f}")
    ax.set_xlabel("Distanza normalizzata dal centroide (RMS z-score)")
    ax.set_ylabel("Frequenza (N osservazioni Monte Carlo)")
    ax.set_title("Densità delle osservazioni in funzione della distanza dal centroide\n"
                 f"Escluse {n_excluded}/{len(distances_full)} osservazioni oltre il cut-off "
                 f"({n_excluded/len(distances_full)*100:.1f}%)")
    ax.legend(fontsize=8)
    fig.tight_layout()
    fig.savefig(chart_path, dpi=150)

    plt.show()
    plt.close(fig)

    return distances_kept, threshold


def make_download_buttons_c4(file_specs: List[Tuple[str, str]]) -> widgets.VBox:
    rows = []
    for path, label in file_specs:
        btn = widgets.Button(description=f"⬇️ Scarica: {label}", button_style="")
        btn_out = widgets.Output()

        def _make_handler(p=path, o=btn_out):
            def _handler(_b):
                with o:
                    clear_output()
                    if IN_COLAB:
                        files.download(p)
                        print(f"Download avviato: {p}")
                    else:
                        print("Download automatico disponibile solo in Google Colab.")
            return _handler

        btn.on_click(_make_handler())
        rows.append(widgets.VBox([btn, btn_out]))

    return widgets.VBox(rows)


# ============================================================
# UI — solo se i prerequisiti sono soddisfatti
# ============================================================

if _ready:

    title = widgets.HTML("<h3>Gestione del Target Probabilistico</h3>")

    origin_names = state["origin_names"]
    var_cols = state["var_cols"]

    minmax_title = widgets.HTML(
        "<b>Limiti Min. % / Max. % per origine</b><br>"
        "Valori pre-impostati come da tabella di riferimento MixManager — modificabili."
    )
    header_row = widgets.HBox([
        widgets.Label("Origine", layout=widgets.Layout(width="220px")),
        widgets.Label("Min. %", layout=widgets.Layout(width="90px")),
        widgets.Label("Max. %", layout=widgets.Layout(width="90px")),
    ])

    min_widgets_prob: Dict[str, widgets.BoundedFloatText] = {}
    max_widgets_prob: Dict[str, widgets.BoundedFloatText] = {}
    minmax_rows = [header_row]

    for name in origin_names:
        default_min, default_max = get_default_minmax(name)  # definita in Cella 1
        min_w = widgets.BoundedFloatText(value=default_min, min=0.0, max=100.0, step=0.5,
                                          layout=widgets.Layout(width="90px"))
        max_w = widgets.BoundedFloatText(value=default_max, min=0.0, max=100.0, step=0.5,
                                          layout=widgets.Layout(width="90px"))
        min_widgets_prob[name] = min_w
        max_widgets_prob[name] = max_w
        minmax_rows.append(widgets.HBox([
            widgets.Label(name, layout=widgets.Layout(width="220px")), min_w, max_w
        ]))

    minmax_table = widgets.VBox(minmax_rows, layout=widgets.Layout(border="1px solid #ddd", padding="6px"))

    load_prob_button = widgets.Button(description="Carica Target probabilistico (.xlsx)", button_style="info")
    prob_out = widgets.Output()
    prob_download_out = widgets.Output()

    display(widgets.VBox([
        title, minmax_title, minmax_table, widgets.HTML("<hr>"),
        load_prob_button, prob_out, prob_download_out
    ]))

    # stato locale della cella, usato dal pulsante di refresh del cut-off
    _c4_state: Dict = {}

    def on_load_prob_clicked(_):
        with prob_download_out:
            clear_output()
        with prob_out:
            clear_output()

            print("--- Limiti Min/Max % per origine (valori impostati in questa cella) ---")
            minmax_current = {}
            for name in origin_names:
                mn = float(min_widgets_prob[name].value)
                mx = float(max_widgets_prob[name].value)
                if mn > mx:
                    mn, mx = mx, mn
                minmax_current[name] = (mn, mx)
                print(f"  {name}: {mn:.1f}–{mx:.1f}%")

            try:
                raw_df = load_probabilistic_target_from_upload()
            except Exception as e:
                print(f"❌ {e}")
                return

            target_stats_df, format_label = build_target_probabilistico_stats(raw_df)
            print(f"\nℹ️ Formato riconosciuto: {format_label}")
            print(f"✅ Statistiche calcolate per {target_stats_df.shape[0]} variabili "
                  f"(mediana, p05, p95, sigma).")

            report = align_target_probabilistico_columns(var_cols, target_stats_df)
            print("\n--- Allineamento colonne (compound) tra dataset generato e Target probabilistico ---")
            print(f"  Solo nel dataset generato (Cella 1): {report['n_only_generated']}")
            if report["only_generated"]:
                print(f"    {report['only_generated']}")
            print(f"  Solo nel Target probabilistico caricato: {report['n_only_target']}")
            if report["only_target"]:
                print(f"    {report['only_target']}")
            print(f"  Variabili comuni: {report['n_common']}")

            print("\n--- Anteprima statistiche Target probabilistico (prime 10 variabili) ---")
            display(target_stats_df.head(10).round(4))

            # --- Metriche di densità + istogramma con cut-off, solo se matrice grezza ---
            is_raw_matrix = format_label.startswith("matrice grezza")
            distance_chart_path = "distanza_dal_centroide_target_probabilistico.png"

            if is_raw_matrix:
                numeric_df = raw_df.apply(pd.to_numeric, errors="coerce")
                numeric_df = numeric_df.loc[:, numeric_df.notna().any(axis=0)]
                distances_full = compute_raw_distances(numeric_df, target_stats_df)

                # --- controlli cut-off percentile + refresh ---
                cutoff_widget = widgets.BoundedFloatText(
                    value=CUTOFF_PERCENTILE_DEFAULT, min=1.0, max=100.0, step=0.5,
                    description="Cut-off percentile:", style={"description_width": "initial"}
                )
                refresh_button = widgets.Button(description="🔄 Aggiorna grafico", button_style="warning")
                stats_out = widgets.Output()
                hist_render_out = widgets.Output()

                def _refresh_histogram(_=None):
                    with hist_render_out:
                        clear_output(wait=True)
                        distances_kept, threshold = render_centroid_distance_histogram(
                            distances_full, float(cutoff_widget.value), distance_chart_path
                        )
                    with stats_out:
                        clear_output(wait=True)
                        distance_stats_full = compute_centroid_distance_stats(distances_full)
                        distance_stats_kept = compute_centroid_distance_stats(distances_kept)
                        print(f"--- Metriche di densità: distanza dal centroide (soglia P{cutoff_widget.value:g} "
                              f"= {threshold:.4f}) ---")
                        comparison = pd.DataFrame([distance_stats_full, distance_stats_kept],
                                                   index=["Tutte le osservazioni", "Dopo cut-off"])
                        display(comparison.round(4))

                    _c4_state["distances_full"] = distances_full
                    _c4_state["distances_kept"] = distances_kept
                    _c4_state["cutoff_percentile"] = float(cutoff_widget.value)
                    _c4_state["cutoff_threshold"] = threshold
                    _c4_state["distance_stats_full"] = compute_centroid_distance_stats(distances_full)
                    _c4_state["distance_stats_kept"] = compute_centroid_distance_stats(distances_kept)

                refresh_button.on_click(_refresh_histogram)

                print("\n--- Istogramma: densità vs distanza dal centroide, con cut-off regolabile ---")
                display(widgets.VBox([cutoff_widget, refresh_button, hist_render_out, stats_out]))
                _refresh_histogram()  # prima visualizzazione, con il default P95

            else:
                print("\nℹ️ Il file è nel formato 'Monte Carlo classico' (solo mediana/p05/p95/sigma, "
                      "senza le osservazioni individuali): le metriche di densità e l'istogramma "
                      "distanza-dal-centroide richiedono le osservazioni grezze e qui non sono "
                      "calcolabili. Carica la matrice grezza multi-riga se ti servono.")
                _c4_state.clear()

            # --- Salvataggio output ---
            target_stats_xlsx = "Target_probabilistico_statistiche.xlsx"
            with pd.ExcelWriter(target_stats_xlsx) as writer:
                target_stats_df.to_excel(writer, sheet_name="Statistiche_per_variabile", index=False)
                pd.DataFrame({
                    "Origine": list(minmax_current.keys()),
                    "Min_%": [v[0] for v in minmax_current.values()],
                    "Max_%": [v[1] for v in minmax_current.values()],
                }).to_excel(writer, sheet_name="Limiti_MinMax", index=False)
                if "distance_stats_full" in _c4_state:
                    pd.DataFrame([_c4_state["distance_stats_full"]]).to_excel(
                        writer, sheet_name="Distanza_tutte_osservazioni", index=False)
                    pd.DataFrame([_c4_state["distance_stats_kept"]]).to_excel(
                        writer, sheet_name="Distanza_dopo_cutoff", index=False)

            print(f"\n✅ Creato {target_stats_xlsx} (Statistiche_per_variabile, Limiti_MinMax"
                  + (", Distanza_tutte_osservazioni, Distanza_dopo_cutoff" if "distance_stats_full" in _c4_state else "")
                  + ").")

            download_specs = [(target_stats_xlsx, "statistiche Target probabilistico + limiti Min/Max")]
            if is_raw_matrix:
                download_specs.append((distance_chart_path, "istogramma distanza dal centroide (con cut-off)"))

            with prob_download_out:
                display(widgets.HTML("<hr><b>Download file generati:</b> "
                                      "(il grafico riflette sempre l'ultimo cut-off applicato)"))
                display(make_download_buttons_c4(download_specs))

            state["target_probabilistico"] = {
                "stats_df": target_stats_df,
                "format_label": format_label,
                "alignment_report": report,
                "minmax_limits": minmax_current,
                **_c4_state,
            }
            print("\n✅ Target probabilistico, limiti Min/Max e metriche salvati in memoria "
                  "(state['target_probabilistico']).")

    load_prob_button.on_click(on_load_prob_clicked)
```
